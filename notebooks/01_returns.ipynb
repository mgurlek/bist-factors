```python
squares = []
for i in range(1, 11):
    squares.append(i ** 2)
print(squares)


def mean(values):
    total = sum(values)
    count = len(values)
    return total / count


print(mean([2, 4, 6, 8]))  
print(mean(squares))       
```

```python
returns = [0.02] * 12
growth = 1.0
for r in returns:
    growth = growth * (1 + r)
print(growth)
```

```python
prices = pd.Series([100, 120, 108, 129.6])
returns = prices.pct_change()
print(returns)
```

```python
returns1 =[]
for i in range(1, len(prices)):
    returns1.append(prices[i]/prices[i-1]-1)

print(returns1)
```

```python
multipliers = returns + 1

total_return = multipliers.prod() - 1
print(total_return)  


log_returns = np.log(multipliers)
print(np.exp(log_returns.sum()) - 1)
```
